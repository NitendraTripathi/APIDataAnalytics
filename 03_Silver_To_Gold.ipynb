```python
# DBTITLE 1,Importing libraries
from pyspark.sql.functions import *
```

```python
silver_df=spark.table('crickapidata.silver.cricket_silver_current_matches')
display(silver_df)
```

```python
# DBTITLE 1,GOLD Analytics 1 :Match type Distribution
gold_match_type_df=silver_df.groupBy('match_type').agg(count('*').alias('Total_Matches'))
display(gold_match_type_df)
```

```python
gold_match_type_df.write\
    .format('delta')\
    .mode('overwrite')\
    .option("overwriteSchema", "true")\
    .saveAsTable('crickapidata.gold.cricket_gold_match_type')
```

```python
gold_venue_df=silver_df.groupBy('venue').agg(count('*').alias('Total_Matches'))
display(gold_venue_df)
```

```python
gold_venue_df.write\
    .format('delta')\
    .mode('overwrite')\
    .option("overwriteSchema", "true")\
    .saveAsTable('crickapidata.gold.gold_venue_df')
```
